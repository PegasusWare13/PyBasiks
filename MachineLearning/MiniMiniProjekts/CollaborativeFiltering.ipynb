```python
from scipy import spatial
user_a = [1,1]
user_b = [2,2]
user_c = [5,3]
user_d = [2,4]

print(spatial.distance.euclidean(user_a,user_b))
print(spatial.distance.euclidean(user_b,user_c))
print(spatial.distance.euclidean(user_c,user_d))
print(spatial.distance.euclidean(user_c,user_a))


print(spatial.distance.cosine(user_a,user_b))
print(spatial.distance.cosine(user_b,user_c))
print(spatial.distance.cosine(user_c,user_d))
print(spatial.distance.cosine(user_d,user_a))
```

```python
from surprise import Dataset, SVD, Reader
from surprise.model_selection import GridSearchCV
reader  = Reader(
    line_format = "user item rating timestamp",
    sep = '\t',
    rating_scale = (1,5) #Inclusive
)
data = Dataset.load_from_file(r'C:\Users\Anwesh\Desktop\Py\MachineLearning\DataBase\ml-100k\u.data', reader=reader)
algorithm = SVD() #Singular Value Decomposition 
parameter_grid = {
    'n_epochs' : [1,20], #all these parameters have to be iterables
    'n_factors' : [80,250], 
    'lr_all' : [0.0001,0.0010],
    "reg_all": [0.3,0.8]
}
GS = GridSearchCV(SVD, param_grid=parameter_grid, cv = 5, measures=["rmse", "mse"])
GS.fit(data)
print(GS.best_params['rmse'])
```
